```python
import pandas as pd
import numpy as np
import joblib
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# Load preprocessed sample evaluation set & trained model
model = joblib.load("models/decision_tree_model.joblib")
X_sample = pd.read_csv("data/X_test_sample.csv")
y_sample = pd.read_csv("data/y_test_sample.csv").values.ravel()
print(f"Loaded model: {model}")
print(f"Sample evaluation shape: {X_sample.shape}")
```

```python
# Evaluate Sample Predictions
preds = model.predict(X_sample)
print(classification_report(y_sample, preds, target_names=["No Subscription", "Subscribed"]))
```

```python
# Explain Predictions with TreeSHAP
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_sample)
if isinstance(shap_values, list):
    sv = shap_values[1]
elif len(shap_values.shape) == 3:
    sv = shap_values[:, :, 1]
else:
    sv = shap_values
shap.summary_plot(sv, X_sample, plot_type="bar", show=False)
plt.tight_layout()
plt.show()
```
